```python
#  replacing null values with median 
data['Review Rating']= data.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
data.isnull().sum()
```

```python
data.columns=data.columns.str.lower()
data.columns=data.columns.str.replace(" ",'_')
data=data.rename(columns={"purchase_amount_(usd)":'purchase_amount'})
```

```python
data.columns
```

```python
# create column age_group
# q cut is used to divide numerical data into groups based on quantiles
labels=['young','adult','middle-aged','senior']
data['age_grp']=pd.qcut(data['age'],q=4,labels=labels)
```

```python
data[['age','age_grp']].head(10)
```

```python
# create column purchase frequency days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
data['purchase_days']=data['frequency_of_purchases'].map(frequency_mapping)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=data,x='gender')

plt.title('Customer Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data['age'], bins=20, kde=True)

plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.countplot(data=data, x='category')
plt.title('Customer Count by Product Category')
plt.xlabel('Category')
plt.ylabel('Number of Customers')
plt.show()
```

```python
season_count = data['season'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    season_count,
    labels=season_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Seasonal Shopping Distribution')
plt.show()
```

```python
plt.figure(figsize=(12,5))
sns.countplot(data=data, x='item_purchased',
              order=data['item_purchased'].value_counts().index)
plt.title('Most Purchased Items')
plt.xlabel('Item Purchased')
plt.ylabel('Number of Purchases')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.countplot(data=data, x='shipping_type',
              order=data['shipping_type'].value_counts().index)

plt.title('Customer Shipping Preferences')
plt.xlabel('Shipping Type')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.show()
# data.columns
```

```python
# 1. Total Customers
total_customers = data['customer_id'].nunique()

# 2. Total Revenue
total_revenue = data['purchase_amount'].sum()

# 3. Average Purchase Amount
avg_purchase = data['purchase_amount'].mean()

# 4. Average Review Rating
avg_rating = data['review_rating'].mean()

# 5. Subscription Rate
subscription_rate = (
    data['subscription_status'].eq('Yes').mean() * 100
)

# 6. Discount Usage Rate
discount_rate = (
    data['discount_applied'].eq('Yes').mean() * 100
)

# 7. Average Previous Purchases
avg_previous_purchases = data['previous_purchases'].mean()

# 8. Average Purchase Frequency
avg_frequency = data['frequency_of_purchases'].mode()[0]

# Display KPIs
print("TOTAL CUSTOMERS:", total_customers)
print("TOTAL REVENUE:", round(total_revenue, 2))
print("AVERAGE PURCHASE AMOUNT:", round(avg_purchase, 2))
print("AVERAGE REVIEW RATING:", round(avg_rating, 2))
print("SUBSCRIPTION RATE:", round(subscription_rate, 2), "%")
print("DISCOUNT USAGE RATE:", round(discount_rate, 2), "%")
print("AVERAGE PREVIOUS PURCHASES:", round(avg_previous_purchases, 2))
print("MOST COMMON PURCHASE FREQUENCY:", avg_frequency)
```
